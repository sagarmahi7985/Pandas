### PANDAS SERIES
A series is a one-dimensional labeled array capable of holding any data type. The axis labels are collectively called the index

### Creating Series 
There are multiple ways to create pandas series

```python
import numpy as np
import pandas as pd
```

```python
labels = ['a', 'b', 'c']
my_list = [10, 20, 30]
arr = np.array([10, 20, 30])
d = {1:10, 2:20, 3:30}
```

```python
pd.Series(my_list)
```

```python
pd.Series(my_list, index=labels)
```

```python
pd.Series(arr)
```

```python
pd.Series(d)
```

### Creating a dataFrame

```python
data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 30, 29, 42],
    'City': ['New York', 'Berlin', 'London', 'Paris'],
    'Salary': [65000, 70000, 62000, 850000]
}

df = pd.DataFrame(data)
df
```

```python
data_list = [
  ['John', 'Anna', 'Peter', 'Linda'],
 [28, 30, 29, 42],
 ['New York', 'Berlin', 'London', 'Paris'],
 [65000, 70000, 62000, 850000]
]
df2 =pd.DataFrame(data_list)
df2
```

```python
data_list = [
 ['John', 28, 'New York', 65000],
 ['Anna', 32, 'Paris', 70000],
 ['Peter', 29, 'Berlin', 62000],
 ['Linda', 42, 'London', 85000]
]
df2 =pd.DataFrame(data_list)
columns = ["Name", "Age", "City", "Salary"]
df2 = pd.DataFrame(data_list, columns=columns)
df2
```

```python
df2
```

#### Selection and Indexing of Columns

```python
df2[["Name", "City"]]
```

#### conditional selection

```python
df2
```

```python
# i only want to see those people whose age is above 30
df2[df2["Age"] > 30]
```

```python
# i only want to see those people whose age is above 30 and their city should be paris
df2[(df2["Age"] > 30) & (df2["City"] == 'Paris')]
```
